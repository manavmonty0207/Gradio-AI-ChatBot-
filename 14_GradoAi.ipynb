```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr

load_dotenv()
# BASE_URL = "https://api.groq.com/openai/v1"
# api_key = os.getenv('GROQ_API_KEY')
# MODEL = 'openai/gpt-oss-120b'


BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
api_key = os.getenv('GOOGLE_API_KEY')
MODEL = 'gemini-3.6-flash'

myAI = OpenAI(base_url=BASE_URL, api_key=api_key)

# Function called by Gradio
def ask_ai(prompt):
    response = myAI.chat.completions.create(
        model = MODEL,
        messages=[
            {"role": "system" , "content": "Give the Answer within 50 Words." },
            {"role": "user" , "content" : prompt}
        ]
    )
    return response.choices[0].message.content

#Gradio Interface

demo = gr.Interface(
    fn = ask_ai,
    inputs = gr.Textbox(
        lines = 5,
        placeholder = "Ask me anything.."
    ),
    outputs = gr.Textbox(
        lines = 10,
        label = "AI Response"
    ),
    title="My First AI ChatBot"
)

#demo.launch()
demo.launch(inbrowser=True , inline = False)
```
